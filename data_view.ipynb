```python
import pickle

# Đường dẫn đến file pkl của bạn
file_path = 'output/interaction_matrix.pkl'

# Mở file ở chế độ đọc nhị phân ('rb')
with open(file_path, 'rb') as f:
    data = pickle.load(f)

# Kiểm tra kiểu dữ liệu và nội dung
print("Kiểu dữ liệu:", type(data))
print(f"Nội dung dữ liệu:\n{data}")
```
