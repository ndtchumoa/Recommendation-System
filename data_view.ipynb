```python
import pandas as pd

# Đường dẫn đến file mapping
file_path = 'output/item_mapping.pkl'

# Đọc file
mapping = pd.read_pickle(file_path)

print("Kiểu dữ liệu gốc:", type(mapping))
print(f"Tổng số sản phẩm có trong dữ liệu: {len(mapping)}")

# Ép kiểu sang DataFrame để có giao diện bảng đẹp mắt
view_df = pd.Series(mapping).to_frame(name="Tên Sản Phẩm")
view_df.index.name = "Mã Sản Phẩm (ID)"

with pd.option_context('display.max_rows', None):
    display(view_df)
```
